```python
from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"
```

```python

# DICTIONARY 1: UNIVERSITY INFORMATION

university_data = {

    "University Name": "Federal Urdu University of Arts, Science & Technology (FUUAST)",

    "Department": "Computer Science",

    "Program": "BS Computer Science",

    "Semester": "6B",

    "Duration": "4 Years",

    "Total Semesters": 8
}



# DICTIONARY 2: COURSES AND TEACHERS


course_data = {

    "Machine Learning": "Dr. Hassan",

    "Database Management System": "Ms. Areeba",

    "Software Engineering": "Dr. Salman",

    "Data Structures": "Ms. Hira"
}


# DICTIONARY 3: STUDENT INFORMATION


student_data = {

    "Muskan Shahid": {
        "Semester": 6,
        "CGPA": 3.62,
        "Performance": "Excellent"
    },

    "Ayesha Khan": {
        "Semester": 6,
        "CGPA": 3.45,
        "Performance": "Very Good"
    },

    "Sara Ahmed": {
        "Semester": 5,
        "CGPA": 3.20,
        "Performance": "Good"
    }
}

# UNIVERSITY AGENT


def university_agent(user_input):

    query = user_input.lower().strip()


    # UNIVERSITY INFORMATION


    if "university name" in query or "name of university" in query:

        return university_data["University Name"]


    elif "department" in query:

        return university_data["Department"]


    elif "program" in query or "degree" in query:

        return university_data["Program"]


    elif "duration" in query or "how many years" in query:

        return university_data["Duration"]


    elif "how many semesters" in query or "total semesters" in query:

        return f"There are {university_data['Total Semesters']} semesters."


    # COURSE / TEACHER INFORMATION


    elif "teacher" in query or "teach" in query or "teaches" in query:

        # Machine Learning
        if "machine learning" in query:

            return f"Machine Learning is taught by {course_data['Machine Learning']}."


        # Database Management System
        elif "database" in query or "dbms" in query:

            return f"Database Management System is taught by {course_data['Database Management System']}."


        # Software Engineering
        elif "software engineering" in query:

            return f"Software Engineering is taught by {course_data['Software Engineering']}."


        # Data Structures
        elif "data structures" in query or "data structure" in query:

            return f"Data Structures is taught by {course_data['Data Structures']}."


        # All courses
        else:

            return """
BSCS Semester 6B Courses and Teachers:

Machine Learning: Dr. Hassan
Database Management System: Ms. Areeba
Software Engineering: Dr. Salman
Data Structures: Ms. Hira
"""

    # COURSE INFORMATION

    elif "machine learning" in query:

        return f"Machine Learning Teacher: {course_data['Machine Learning']}"


    elif "database" in query or "dbms" in query:

        return f"Database Management System Teacher: {course_data['Database Management System']}"


    elif "software engineering" in query:

        return f"Software Engineering Teacher: {course_data['Software Engineering']}"


    elif "data structures" in query or "data structure" in query:

        return f"Data Structures Teacher: {course_data['Data Structures']}"

    # STUDENT INFORMATION

    for student_name, data in student_data.items():

        if student_name.lower() in query:

            if "cgpa" in query:

                return f"{student_name}'s CGPA is {data['CGPA']}."


            elif "semester" in query:

                return f"{student_name} is in semester {data['Semester']}."


            elif "performance" in query:

                return f"{student_name}'s performance is {data['Performance']}."


            else:

                return f"""
Student: {student_name}
Semester: {data['Semester']}
CGPA: {data['CGPA']}
Performance: {data['Performance']}
"""

    general_questions = [

        "what is",
        "what are",
        "explain",
        "define",
        "meaning of",
        "difference between",
        "how does",
        "why",
        "tell me about"

    ]


    if any(question in query for question in general_questions):

        prompt = f"""

You are FUUAST Smart Assistant.

You help BS Computer Science students.

University:
Federal Urdu University of Arts, Science & Technology (FUUAST)

Program:
BS Computer Science

Answer general educational questions in a simple and clear way.

Topics can include:

- Machine Learning
- Database Management System
- Software Engineering
- Data Structures
- Python
- Programming
- Data Science
- Artificial Intelligence
- Natural Language Processing
- Computer Science
- Algorithms

IMPORTANT:

Do not invent specific teacher information.

Teacher information is stored in Python dictionaries.

Student Question:

{user_input}

Give a simple and understandable answer.

"""

        response = client.models.generate_content(

            model=MODEL,

            contents=prompt

        )

        return response.text


    return "Sorry, this information is not available."

print("FUUAST Smart Assistant Started")

print("Ask me about university, courses, teachers or students.")

print("Type 'exit' to stop.\n")




while True:

    user = input("You: ")


    if user.lower().strip() == "exit":

        print("Goodbye!")

        break


    try:

        reply = university_agent(user)

        print("\nFUUAST Smart Assistant:")

        print(reply)

        print()


    except Exception as e:

        print("Error:", e)
```
