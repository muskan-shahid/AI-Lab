```python

# Travel Agent
# Predefined Rule-Based Agent
# Flights from Karachi to Lahore, Islamabad, and Multan

# Take input and make city names case-insensitive
departure_city = input("Enter Departure City: ").strip().title()
destination_city = input("Enter Destination City: ").strip().title()


# Karachi to Lahore
if departure_city == "Karachi" and destination_city == "Lahore":

    print("\nI've found three available flights for you.")
    budget = int(input("Enter Your Budget in Rupees: "))

    if budget < 15000:
        print("Sorry, with your budget there is no flight available.")

    elif budget <= 20000:
        print("There's a flight available for you.")
        print("Departure Time: 10 PM")
        print("Arrival Time: 3 AM")
        print("Estimated Fare: Rs. 20,000")

    elif budget <= 30000:
        print("There are two flights available for you.")

        print("\n1. Estimated Fare: Rs. 30,000")
        print("   Departure Time: 6 PM")
        print("   Arrival Time: 9 PM")

        print("\n2. Estimated Fare: Rs. 20,000")
        print("   Departure Time: 10 PM")
        print("   Arrival Time: 3 AM")

    elif budget <= 40000:
        print("There are three flights available for you.")

        print("\n1. Estimated Fare: Rs. 36,000")
        print("   Departure Time: 4 PM")
        print("   Arrival Time: 6 PM")

        print("\n2. Estimated Fare: Rs. 28,000")
        print("   Departure Time: 6 PM")
        print("   Arrival Time: 9 PM")

        print("\n3. Estimated Fare: Rs. 20,000")
        print("   Departure Time: 10 PM")
        print("   Arrival Time: 3 AM")

    else:
        print("There are no flights available within this budget range.")


# Karachi to Islamabad
elif departure_city == "Karachi" and destination_city == "Islamabad":

    print("\nI've found two available flights for you.")
    budget = int(input("Enter Your Budget in Rupees: "))

    if budget < 20000:
        print("Sorry, with your budget there is no flight available.")

    elif budget <= 30000:
        print("There's a flight available for you.")
        print("Estimated Fare: Rs. 25,000")
        print("Departure Time: 7 PM")
        print("Arrival Time: 12 AM")

    else:
        print("There are two flights available for you.")

        print("\n1. Estimated Fare: Rs. 34,000")
        print("   Departure Time: 5 PM")
        print("   Arrival Time: 6:30 PM")

        print("\n2. Estimated Fare: Rs. 25,000")
        print("   Departure Time: 7 PM")
        print("   Arrival Time: 12 AM")


# Karachi to Multan
elif departure_city == "Karachi" and destination_city == "Multan":

    print("\nI've found one available flight for you.")
    budget = int(input("Enter Your Budget in Rupees: "))

    if budget < 15000:
        print("Sorry, with your budget there is no flight available.")

    elif budget <= 20000:
        print("There's a flight available for you.")
        print("Estimated Fare: Rs. 19,000")
        print("Departure Time: 5:30 PM")
        print("Arrival Time: 8:30 PM")

    else:
        print("Sorry, there are no available flights within this budget range.")


# Invalid route
else:
    print(
        "\nSorry, there are no available flights from",
        departure_city,
        "to",
        destination_city,
        "!"
    )

```
