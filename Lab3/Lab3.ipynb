```python
from google import genai
client = genai.Client(api_key="")
```

```python
MODEL = "models/gemini-3.5-flash"


def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
def code_sentinel(user_input):

    prompt = f"""
You are CodeSentinel AI, an intelligent Python Security and
Refactoring Agent.

Your job is to analyze Python code and perform four main tasks:

1. DETECT
   Identify security vulnerabilities, bugs, unsafe functions,
   and poor coding practices.

2. EXPLAIN
   Explain each problem clearly and simply.
   Mention why the problem is dangerous or incorrect.

3. FIX
   Provide a safer and improved version of the Python code.
   Preserve the original functionality whenever possible.

4. VERIFY
   Check the corrected code again and explain whether the
   identified problems have been fixed.

Look especially for:
- Command injection
- SQL injection
- Hardcoded passwords or API keys
- Unsafe subprocess usage
- Dangerous use of eval() or exec()
- Unsafe file handling
- Insecure input handling
- Weak or unsafe coding practices

Use the following response format:

### Security Analysis
List the detected vulnerabilities and coding problems.

### Explanation
Explain why each problem is dangerous or incorrect.

### Secure Code
Provide the corrected Python code.

### Verification
Explain whether the security issues have been fixed.

Important Rules:
- Do not invent vulnerabilities.
- Only report problems supported by the provided code.
- Keep explanations simple and understandable.
- Preserve the original purpose of the code.
- Never expose passwords, API keys, or other secrets.

User Code / Request:
{user_input}

CodeSentinel AI Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("CodeSentinel AI Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = code_sentinel(user)
        print("\nCodeSentinel AI:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
